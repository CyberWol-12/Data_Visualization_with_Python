```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/Canada.csv")
print("data downloded")
```

```python
df.index.name = None
```

```python
years = list(map(str,range(1980, 2014)))
years
```

```python
haiti = df.loc['Haiti',years]
haiti
```

```python
haiti.head()
```

```python
haiti.plot()
```

```python
haiti.plot(kind='line')

plt.title('Immigration from Haiti')
plt.ylabel('Number of immigrants')
plt.xlabel('Years')

plt.show()
```

```python
haiti.index = haiti.index.map(int)
haiti.plot(kind = 'line')

plt.title('Immigration from Haiti')
plt.ylabel('Number of immigrants')
plt.xlabel('Years')
# annotate the 2010 Earthquake. 
# syntax: plt.text(x, y, label)
plt.text(2000,6000,'2010 Earthquake')
plt.show()
```

```python
years = list(map(str,range(1980, 2014)))
years
```

```python
df.sort_values(['Total'],ascending=False,axis = 0,inplace=True)
top5 = df.head()
top5 = top5[years].transpose()
top5
```

```python
top5.index = top5.index.map(int)
top5.plot(kind = 'area',
          stacked=False,
          figsize=(20,10))
plt.title('Immigration Trend of Top 5 Countries')
plt.xlabel('Number of Immigrants')
plt.ylabel('Years')
plt.show()
```

```python
ax  = top5.plot(kind='area',alpha = 0.35,figsize=(20,10))
ax.set_title('Immigration Trend of Top 5 Countries')
ax.set_ylabel('Number of Immigrants')
ax.set_xlabel('Years')
```
